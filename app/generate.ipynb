```python
import pandas as pd
import random

# Sample data within England's latitude and longitude frames
data = {
    'building_id': list(range(1, 5001)),
    'latitude': [random.uniform(49.9, 52.8) for _ in range(5000)],  # England's latitude range
    'longitude': [random.uniform(-3.4, 0) for _ in range(5000)],  # England's longitude range
    'CO2_Usage': [random.randint(1000, 2000) for _ in range(5000)],
    'Water_Usage': [random.randint(4000, 6000) for _ in range(5000)],
    'Energy_Consumption': [random.randint(1500, 2500) for _ in range(5000)]
}

# Create a DataFrame
df = pd.DataFrame(data)

# Save to CSV
df.to_csv('sample_buildings.csv', index=False)

print("CSV file 'sample_buildings.csv' created successfully.")
```
